## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
```

```python
not_updated_90d = df["days_since_last_update"] > 90
visible = df["impressions_90d"] >= 100
older_and_visible = not_updated_90d & visible
print(
    f"{older_and_visible.sum():,} pages ({older_and_visible.mean():.1%} of all {len(df)} pages) "
    "have not been updated in more than 90 days "
    "and received at least 100 impressions."
)
```

```python
visible = df["impressions_90d"] >= 100
older = df["days_since_last_update"] > 90
declining = df["trend_direction"].str.lower().eq("down")

declining_rate_recent_visible = declining[visible & ~older].mean()
declining_rate_older_visible = declining[visible & older].mean()
difference = declining_rate_older_visible - declining_rate_recent_visible

print(
    f"Downward-trending rate among visible pages updated within the last 90 days: "
    f"{declining_rate_recent_visible:.1%}"
)
print(
    f"Downward-trending rate among visible pages older than 90 days: "
    f"{declining_rate_older_visible:.1%}"
)
print(f"Observed difference: {difference:.1%}")
```

Of the 30,000 pages, 8,119 (27.1%) received at least 100 impressions and had not been updated in more than 90 days, creating a sizable pool for editorial review. Among visible pages, 62.3% of those not updated in more than 90 days had a measured downward trend, compared with 58.3% of those updated within 90 days, a difference of 4.0 percentage points. This observed association suggests that time since the last update may be useful as one signal for prioritizing pages for review, but it does not show that page age caused the decline or that refreshing a page would improve traffic.
